```python
import pandas as pd
import matplotlib.pyplot as plt

# Carregando o arquivo usando o caminho correto
df = pd.read_csv("/estudo.csv", parse_dates=["data"])

# Limpeza
df = df.drop_duplicates()
df["materia"] = df["materia"].str.strip().str.title()

# Horas por matéria
por_materia = df.groupby("materia")["horas"].sum().sort_values()
por_materia.plot(kind="barh", title="Horas por matéria")
plt.xlabel("Horas")
plt.tight_layout()
plt.savefig("horas_por_materia.png")
plt.show()

# Horas por semana
df["semana"] = df["data"].dt.to_period("W").astype(str)
por_semana = df.groupby("semana")["horas"].sum()
por_semana.plot(kind="bar", title="Horas por semana")
plt.ylabel("Horas")
plt.tight_layout()
plt.savefig("horas_por_semana.png")
plt.show()

# Dia da semana com mais estudo
df["dia"] = df["data"].dt.day_name()
print(df.groupby("dia")["horas"].mean().sort_values(ascending=False))
```
